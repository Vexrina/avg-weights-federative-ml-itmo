```python
channel = grpc.insecure_channel("localhost:9090")

stub = serverside_pb2_grpc.AvgWeightsStub(channel)
```

```python
state = {
    "w": torch.tensor(1.0)
}

buf = io.BytesIO()
torch.save(state, buf)
weights_bytes = buf.getvalue()

request = serverside_pb2.AddMyWeightsRequest(
    client_id="87297556-df48-45c8-a360-0ed9067f41e1",
    weights=weights_bytes,       # ← ВАЖНО: просто bytes
    num_examples=10,
)

response = stub.AddMyWeights(request)

print("OK, response:", response)
```
